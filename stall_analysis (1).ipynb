# Gikomba Stall Sales Analysis

## PLP Data Analytics Assignment 1

This analysis examines Wanjiku's fresh produce business performance using two weeks of sales and inventory data. The objective is to understand business growth, identify products that need restocking, evaluate profitability, and provide recommendations based on sales trends.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

# Task 1: Sales Statistics

This task calculates total, average, highest and lowest sales and identifies the best and worst performing days.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

highest_index = daily_sales.index(highest_sales)
lowest_index = daily_sales.index(lowest_sales)

best_day = days[highest_index]
worst_day = days[lowest_index]

best_week = 1 if highest_index < 7 else 2
worst_week = 1 if lowest_index < 7 else 2

print("TASK 1")
print(f"Total Sales: KES {total_sales:,.0f}")
print(f"Average Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,.0f}")
print(f"Lowest Sales: KES {lowest_sales:,.0f}")
print(f"Best Day: {best_day} (Week {best_week})")
print(f"Worst Day: {worst_day} (Week {worst_week})")
```

# Task 2: Weekly Totals and Growth

This task uses a function and list slicing to determine weekly totals and growth rate.

```python
def week_totals(sales):
    """
    Returns week 1 and week 2 sales totals.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2


week1_total, week2_total = week_totals(daily_sales)

growth_rate = ((week2_total - week1_total) / week1_total) * 100

print("TASK 2")
print(f"Week 1 Total: KES {week1_total:,.0f}")
print(f"Week 2 Total: KES {week2_total:,.0f}")
print(f"Growth Rate: {growth_rate:.2f}%")
```

# Task 3: Sales Classification

This task classifies each day as Strong, Normal, or Weak.

```python
def classify_day(amount, average):
    """
    Classify a day's sales performance.
    """
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"


categories = {}

for sale in daily_sales:
    result = classify_day(sale, average_sales)

    if result in categories:
        categories[result] += 1
    else:
        categories[result] = 1

print("TASK 3")
print(categories)
```

# Task 4: Weekday vs Weekend Performance

This task compares weekday and weekend sales performance.

```python
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Sat", "Sun"]:
        weekend_sales.append(daily_sales[i])
    else:
        weekday_sales.append(daily_sales[i])

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

difference = ((weekend_average - weekday_average) / weekday_average) * 100

print("TASK 4")
print(f"Weekday Average: KES {weekday_average:,.2f}")
print(f"Weekend Average: KES {weekend_average:,.2f}")
print(f"Difference: {difference:.2f}%")
```

# Task 5: Inventory Analysis

This task identifies products that should be reordered and calculates gross margins.

```python
def reorder_products(stock_data):
    """
    Returns products that need restocking.
    """
    products = []

    for product, details in stock_data.items():
        if details["qty"] <= details["reorder"]:
            products.append(product)

    return products


items_to_reorder = reorder_products(stock)

print("TASK 5")
print("Products to Reorder:")
print(items_to_reorder)

print("\nGross Margins")

for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product}: {margin:.2f}%")
```

# Task 6: Stall Performance Report

This task presents all findings in a professionally formatted report.

```python
print("=" * 60)
print("        STALL PERFORMANCE REPORT")
print("=" * 60)

print(f"Total Sales:              KES {total_sales:,.0f}")
print(f"Average Daily Sales:      KES {average_sales:,.2f}")
print(f"Highest Sales:            KES {highest_sales:,.0f}")
print(f"Lowest Sales:             KES {lowest_sales:,.0f}")

print(f"\nBest Day: {best_day} (Week {best_week})")
print(f"Worst Day: {worst_day} (Week {worst_week})")

print(f"\nWeek 1 Total: KES {week1_total:,.0f}")
print(f"Week 2 Total: KES {week2_total:,.0f}")
print(f"Growth Rate: {growth_rate:.2f}%")

print("\nDay Classification:")
for category, count in categories.items():
    print(f"{category}: {count}")

print("\nProducts To Reorder:")
for product in items_to_reorder:
    print(f"- {product}")

print("\nGross Margins:")
for product, details in stock.items():
    margin = ((details['price'] - details['cost']) / details['price']) * 100
    print(f"{product}: {margin:.2f}%")

print("=" * 60)
```
